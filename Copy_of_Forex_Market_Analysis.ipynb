# Forex Market Data Analysis Using Python in Google Colab

## Objective

The objective of this project is to analyze historical forex market data using Python.
This project helps in understanding market trends, daily returns, moving averages, and volatility through data visualization techniques in Google Colab.

```python
!pip install yfinance
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
pair = "EURUSD=X"

data = yf.download(pair, start="2023-01-01", end="2025-01-01")

print(data.head())
```

```python
plt.figure(figsize=(12,5))
plt.plot(data['Close'], label='EUR/USD Close Price')
plt.title('EUR/USD Forex Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.grid()
plt.show()
```

```python
data['SMA_20'] = data['Close'].rolling(window=20).mean()
data['SMA_50'] = data['Close'].rolling(window=50).mean()
```

```python
plt.figure(figsize=(12,5))

plt.plot(data['Close'], label='Close Price')
plt.plot(data['SMA_20'], label='20-Day SMA')
plt.plot(data['SMA_50'], label='50-Day SMA')

plt.title('Moving Average Analysis')
plt.legend()
plt.grid()
plt.show()
```

```python
data['Daily_Return'] = data['Close'].pct_change()

print(data['Daily_Return'].describe())
```

```python
plt.figure(figsize=(10,5))

data['Daily_Return'].hist(bins=50)

plt.title('Daily Return Distribution')
plt.show()
```

```python
volatility = data['Daily_Return'].std()

print("Market Volatility:", volatility)
```

```python
print("Highest Price:")
print(data['High'].max())

print("Lowest Price:")
print(data['Low'].min())
```
